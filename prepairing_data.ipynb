```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=PyPDFLoader("bentechcalisma.pdf")
pages=loader.load()
```

```python
pages
```

```python
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    length_function=len
)

chunks=text_splitter.split_documents(pages)
```

```python
len(chunks)
```

```python
chunks[1]
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
```

```python

from langchain_community.vectorstores import FAISS
from my_embeddings import BentechEmbeddings








custom_embeddings = BentechEmbeddings(api_key=api_key)


print("Hacı, FAISS motoru çalıştırılıyor...")
vector_store = FAISS.from_documents(chunks, custom_embeddings)


vector_store.save_local("faiss_bentech_index")

print("Hacı işlem tamam! 'faiss_bentech_index' klasörün oluştu.")
```
